```python
import pandas as pd
```

```python
df = pd.read_csv("../data/cleaned_superstore_data.csv",parse_dates=['order_date', 'ship_date'])
```

```python
df.info()
```

# Total profit

```python
df['profit'].sum()
```

# Total sales

```python
df['sales'].sum()
```

# Profit margin

```python
(df['profit'].sum() / df['sales'].sum() )* 100 
```

# Total orders

```python
df['order_id'].nunique()
```

# Total customers

```python
df['customer_id'].nunique()
```

# Average order value

```python
(df['sales'].sum() /df['order_id'].nunique())
```

# Sales and Profit by category

```python
df.groupby(df['category'])[['sales','profit']].sum().sort_values(by ='sales' , ascending = False)
```

# Sales and Profit by region

```python
df.groupby(df['region'])[['sales','profit']].sum().sort_values(by ='sales' , ascending = False)
```

# Yearly Sales Growth

```python
 df.groupby(df['order_date'].dt.year)['sales'].sum()
```
